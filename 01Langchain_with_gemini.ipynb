```python
%%writefile app.py
import streamlit as st
# from langchain_openai import ChatOpenAI
from langchain_google_genai import ChatGoogleGenerativeAI
from google.colab import userdata

# Automatically fetch the API keys from your Colab Secrets/Environment
# try:
#     openai_api_key = userdata.get('OPENAI_API_KEY')
#     if openai_api_key:
#         os.environ["OPENAI_API_KEY"] = openai_api_key
# except Exception:
#     openai_api_key = None

try:
    gemini_api_key = userdata.get('GEMINI_API_KEY')
except Exception:
    gemini_api_key = None

try:
    langchain_api_key = userdata.get('LANGCHAIN_API_KEY')
except Exception:
    langchain_api_key = None

# Set up LangSmith tracing automatically if the key exists
import os
if langchain_api_key:
    os.environ["LANGCHAIN_TRACING_V2"] = "true"
    os.environ["LANGCHAIN_API_KEY"] = langchain_api_key

# Set up the Streamlit UI
st.set_page_config(page_title="LangChain Quickstart", layout="centered")
st.title("🦜🔗 LangChain + Streamlit App")

# Main prompt interface
user_prompt = st.text_area("Ask the AI:", "What are three fun facts about space?")
submit_button = st.button("Submit")

if submit_button:
    if not gemini_api_key:
        st.error("Missing gemini_api_key! Please check your Google Colab Secrets configuration.")
    else:
        try:
          with st.status("Thinking...", expanded=True) as status:
            # with st.spinner("Thinking..."):
                # Initialize the LangChain ChatOpenAI model
                # llm = ChatGoogleGenerativeAI(model="gpt-4o-mini", temperature=0.7, api_key=openai_api_key)

                llm = ChatGoogleGenerativeAI(
                    model="gemini-3.8-flash",
                    temperature=0.7,
                    google_api_key=gemini_api_key
                )

                # Invoke the model and get a response
                response = llm.invoke(user_prompt)

                # Display results
                st.success("Response:")
                st.write(response.content)
        except Exception as e:
            st.error(f"An error occurred: {e}")
```
